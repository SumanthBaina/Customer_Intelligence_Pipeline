```python
# 03_Gold_Analysis_Dash
from pyspark.sql.functions import col, split

storage_account = "rgstoragecustomer"
gold_path = f"abfss://gold@{storage_account}.dfs.core.windows.net/customer_insights"

# Load the AI-enriched data
gold_df = spark.read.format("delta").load(gold_path)

# Split the 'ai_insight' column into Sentiment and Issue for better plotting
final_df = gold_df.withColumn("Sentiment", split(col("ai_insight"), " \| ").getItem(0)) \
                  .withColumn("Issue", split(col("ai_insight"), " \| ").getItem(1))

display(final_df)
```
